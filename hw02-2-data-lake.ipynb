```python
import boto3
from pathlib import Path

s3 = boto3.client("s3")

bucket_name = "dsan6000-eak140"

s3.create_bucket(Bucket=bucket_name)
```

```python
parquet_files = list(Path("data").glob("*.parquet"))

for file in parquet_files:
    s3.upload_file(str(file), bucket_name, f"wikipedia-hourly/{file.name}")
```
